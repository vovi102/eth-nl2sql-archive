# Base model generation, including compact controls

Render prompts locally for B1/B2 and `b3-base`/`b3l-base`. These compact base controls use exactly the messages of B3/B3L and load no adapter.

Requires a T4 GPU, internet, a Kaggle secret `HF_TOKEN` with model access, and the repo attached as `eth-nl2sql-src`, including `runs/<config>/prompts-dev_eval.jsonl`.
Predictions and resource/provenance manifests are written to `runs/<config>/seed-42/`. Re-running validates prompt/model/adapter identity before reusing output. Keep the common prompt files with the seed artifacts when copying results back. Test v2 remains gated while draft.

```python
from huggingface_hub import login
from kaggle_secrets import UserSecretsClient
login(UserSecretsClient().get_secret('HF_TOKEN'))
```

```python
import subprocess

CONFIGS = ['b1', 'b1l', 'b2', 'b2l', 'b3-base', 'b3l-base']
for name in CONFIGS:
    subprocess.run(['eth-nl2sql', 'preflight', '--config', name], check=True)
    subprocess.run(['eth-nl2sql', 'generate', '--config', name, '--split', 'dev_eval', '--seed', '42'], check=True)
```
